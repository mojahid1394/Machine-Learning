```python
import pandas as ps
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from matplotlib import pyplot as plt
```

```python
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.3)
```

```python
len(X_test)
```

**Logistic Regression**

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
lr.score(X_test, y_test)
```

**SVM**

```python
svm = SVC(gamma='auto')
svm.fit(X_train, y_train)
svm.score(X_test, y_test)
```

**Random Forest**

```python
rf = RandomForestClassifier(n_estimators=40)
rf.fit(X_train, y_train)
rf.score(X_test, y_test)
```

**using K-Fold for our digits example**

```python
def get_score(model, X_train, X_test, y_train, y_test):
    model.fit(X_train, y_train)
    return model.score(X_test, y_test)
```

```python
from sklearn.model_selection import StratifiedKFold
folds = StratifiedKFold(n_splits=3)
```

```python
scores_logistic = []
scores_svm = []
scores_rf = []

for train_index, test_index in folds.split(digits.data, digits.target):
    X_train, X_test, y_train, y_test =  \
    digits.data[train_index], digits.data[test_index], \
    digits.target[train_index], digits.target[test_index]

    scores_logistic.append(get_score(LogisticRegression(max_iter=1000), X_train, X_test, y_train, y_test))
    scores_svm.append(get_score(SVC(gamma='auto'), X_train, X_test, y_train, y_test))
    scores_rf.append(get_score(RandomForestClassifier(n_estimators=40), X_train, X_test, y_train, y_test))
```

**Cross Validation Function**

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(LogisticRegression(max_iter=1000), digits.data, digits.target, cv=3)
```

```python
cross_val_score(SVC(gamma='auto'), digits.data, digits.target, cv=3)
```

```python
cross_val_score(RandomForestClassifier(n_estimators=40), digits.data, digits.target, cv=3)
```

**Parameter tunning using k fold cross validation**

```python
scores1 = cross_val_score(RandomForestClassifier(n_estimators=5), digits.data, digits.target, cv=10)
scores1.mean()
```

```python
scores2 = cross_val_score(RandomForestClassifier(n_estimators=20), digits.data, digits.target, cv=10)
np.average(scores2)
```

```python
scores3 = cross_val_score(RandomForestClassifier(n_estimators=30), digits.data, digits.target, cv=10)
scores3.mean()
```

```python
scores4 = cross_val_score(RandomForestClassifier(n_estimators=40), digits.data, digits.target, cv=10)
scores4.mean()
```

Here we used cross_val_score to fine tune our random forest classifier and figured that having around 40 trees in random forest gives best result.
