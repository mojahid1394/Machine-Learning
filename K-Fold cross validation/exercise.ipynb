```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

**logistic regression**

```python
l_scores = cross_val_score(LogisticRegression(max_iter=1000), iris.data, iris.target, cv=10)
l_scores
```

```python
np.average(l_scores)
```

**Decision Tree**

```python
d_scores = cross_val_score(DecisionTreeClassifier(), iris.data, iris.target, cv=10)
d_scores
```

```python
np.average(d_scores)
```

**Support Vector Machine (SVM)**

```python
svm_scores = cross_val_score(SVC(), iris.data, iris.target, cv=10)
svm_scores
```

```python
np.average(svm_scores)
```

**Random Forest**

```python
rf_scores = cross_val_score(RandomForestClassifier(n_estimators=40), iris.data, iris.target, cv=10)
rf_scores
```

```python
np.average(rf_scores)
```
